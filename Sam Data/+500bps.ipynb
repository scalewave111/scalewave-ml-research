```python
import pyarrow.parquet as pq
import pandas as pd
import numpy as np
import random

random.seed(42)

def sample_dataset(path, max_groups=25, rows_per_group=500):
    pf = pq.ParquetFile(path)

    # Select actual feature-value columns
    feature_cols = [
        c for c in pf.schema.names
        if c.endswith("__value")
    ]

    required = ["signed_return"]
    columns = required + feature_cols

    # Sample row groups spread across the file
    group_ids = sorted(random.sample(
        range(pf.num_row_groups),
        min(max_groups, pf.num_row_groups)
    ))

    samples = []

    for group_id in group_ids:
        table = pf.read_row_group(
            group_id,
            columns=columns
        )
        df = table.to_pandas()

        if len(df) > rows_per_group:
            df = df.sample(
                n=rows_per_group,
                random_state=42
            )

        samples.append(df)

    return pd.concat(samples, ignore_index=True), feature_cols


large_path = files["large_moves"]
small_path = files["smaller_moves"]

large_df, large_features = sample_dataset(large_path)
small_df, small_features = sample_dataset(small_path)

# Keep only features shared by both files
features = sorted(set(large_features) & set(small_features))

print("Sampled large-move rows:", len(large_df))
print("Sampled smaller-move rows:", len(small_df))
print("Shared features:", len(features))

# Split the large-move sample by direction
large_up = large_df[large_df["signed_return"] > 500]
large_down = large_df[large_df["signed_return"] < -500]

small_moves = small_df[
    small_df["signed_return"].between(-500, 500)
]

groups = {
    "Large Up": large_up,
    "Large Down": large_down,
    "Smaller Moves": small_moves
}

# Calculate a standardized difference in feature values
def compare_features(group_a, group_b, feature_cols):
    results = []

    for col in feature_cols:
        a = pd.to_numeric(
            group_a[col], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan).dropna()

        b = pd.to_numeric(
            group_b[col], errors="coerce"
        ).replace([np.inf, -np.inf], np.nan).dropna()

        if len(a) < 30 or len(b) < 30:
            continue

        mean_a, mean_b = a.mean(), b.mean()
        std_a, std_b = a.std(), b.std()

        pooled_std = np.sqrt(
            (std_a**2 + std_b**2) / 2
        )

        if pooled_std == 0 or not np.isfinite(pooled_std):
            continue

        results.append({
            "feature": col,
            "family": col.split("__")[1],
            "mean_group_a": mean_a,
            "mean_group_b": mean_b,
            "standardized_difference":
                (mean_a - mean_b) / pooled_std,
            "valid_n_a": len(a),
            "valid_n_b": len(b)
        })

    result = pd.DataFrame(results)

    if not result.empty:
        result["abs_difference"] = (
            result["standardized_difference"].abs()
        )
        result = result.sort_values(
            "abs_difference", ascending=False
        )

    return result


up_vs_small = compare_features(
    large_up, small_moves, features
)

down_vs_small = compare_features(
    large_down, small_moves, features
)

print("\nTOP 20 FEATURES: LARGE UP vs SMALLER MOVES")
display(up_vs_small.head(20))

print("\nTOP 20 FEATURES: LARGE DOWN vs SMALLER MOVES")
display(down_vs_small.head(20))

print("\nFEATURE FAMILY RANKING: LARGE UP vs SMALLER MOVES")
if not up_vs_small.empty:
    display(
        up_vs_small.groupby("family")["abs_difference"]
        .agg(["mean", "median", "count"])
        .sort_values("mean", ascending=False)
    )

print("\nFEATURE FAMILY RANKING: LARGE DOWN vs SMALLER MOVES")
if not down_vs_small.empty:
    display(
        down_vs_small.groupby("family")["abs_difference"]
        .agg(["mean", "median", "count"])
        .sort_values("mean", ascending=False)
    )
```

```python
import numpy as np
import pandas as pd
import pyarrow.parquet as pq

# Use the paths and feature_cols from the previous cell.
target = "signed_return"
time_col = "decision_time"
sample_size = 15000

def load_sample(path, features, n=15000):
    available = pq.read_schema(path).names
    cols = [c for c in [time_col, target] + features if c in available]

    # Read only required columns; then take a sample.
    df = pd.read_parquet(path, columns=cols)
    if len(df) > n:
        df = df.sample(n=n, random_state=42)
    return df

# Each dataset is sampled independently.
large = load_sample(large_path, feature_cols, sample_size)
small = load_sample(small_path, feature_cols, sample_size)

large["move_group"] = np.where(
    large[target] > 500, "Large Up",
    np.where(large[target] < -500, "Large Down", "Other")
)
small["move_group"] = "Smaller Move"

data = pd.concat([large, small], ignore_index=True)
data[target] = pd.to_numeric(data[target], errors="coerce")

# Rank features by absolute correlation with signed future return.
results = []

for col in feature_cols:
    if col not in data.columns:
        continue

    x = pd.to_numeric(data[col], errors="coerce")
    valid = x.notna() & data[target].notna()

    if valid.sum() < 100:
        continue

    corr = x[valid].corr(data.loc[valid, target], method="spearman")

    results.append({
        "feature": col,
        "spearman_corr": corr,
        "abs_corr": abs(corr) if pd.notna(corr) else np.nan,
        "valid_rows": int(valid.sum())
    })

ranking = (
    pd.DataFrame(results)
    .dropna(subset=["abs_corr"])
    .sort_values("abs_corr", ascending=False)
)

print("TOP 20 FEATURES ASSOCIATED WITH FUTURE RETURN")
display(ranking.head(20))
```

```python
import gc
import numpy as np
import pandas as pd
import pyarrow.parquet as pq
import matplotlib.pyplot as plt

# Explicitly defining the paths to resolve the NameError
large_path = (
    "/content/drive/MyDrive/Scalewave/"
    "BTCUSDT_signed_return_above_500_or_below_minus500_"
    "20261027_093028_e1453352.parquet"
)

small_path = (
    "/content/drive/MyDrive/Scalewave/"
    "BTCUSDT_signed_return_between_minus500_and_plus500_"
    "20261027_101854_1aefb791.parquet"
)

TARGET = "signed_return"
TIME_COL = "decision_time"
MAX_ROWS = 2000
BATCH_SIZE = 128
ROWS_PER_BATCH = 2

# --------------------------------------------------
# 1. Choose a small set of useful feature columns
# --------------------------------------------------
keywords = [
    "basis_persistence",
    "basis_level",
    "liq_side_pressure",
    "liq_activity_severity",
    "cross_asset_liq_activity",
    "flow_pressure",
    "flow_cvd_slope",
    "flow_confirmation",
    "oi_leverage_unwind",
    "funding_level",
    "realized_variance",
    "price_acceleration",
    "price_excursion_range",
    "volatility",
]

schema = pq.read_schema(large_path)
small_schema = pq.read_schema(small_path)

common_cols = set(schema.names) & set(small_schema.names)

feature_cols = [
    c for c in schema.names
    if c in common_cols
    and c.endswith("__value")
    and any(k in c.lower() for k in keywords)
]

# Limit the analysis to 20 columns to reduce memory use.
feature_cols = feature_cols[:20]

print("Selected features:", len(feature_cols))
for c in feature_cols:
    print(c)

# --------------------------------------------------
# 2. Read tiny samples batch-by-batch
# --------------------------------------------------
def read_small_sample(path, feature_cols, max_rows=2000):
    parquet = pq.ParquetFile(path)

    available = set(parquet.schema_arrow.names)
    columns = [
        c for c in [TARGET, TIME_COL] + feature_cols
        if c in available
    ]

    samples = []
    rng = np.random.default_rng(42)

    for batch in parquet.iter_batches(
        batch_size=BATCH_SIZE,
        columns=columns,
    ):
        batch_df = batch.to_pandas()

        # Retain only a few rows from each batch.
        n = min(ROWS_PER_BATCH, len(batch_df))
        if n:
            samples.append(
                batch_df.iloc[
                    rng.choice(len(batch_df), size=n, replace=False)
                ]
            )

        del batch_df, batch

    if not samples:
        return pd.DataFrame(columns=columns)

    result = pd.concat(samples, ignore_index=True)

    if len(result) > max_rows:
        result = result.sample(
            n=max_rows, random_state=42
        ).reset_index(drop=True)

    del samples
    gc.collect()
    return result


large_sample = read_small_sample(large_path, feature_cols, MAX_ROWS)
small_sample = read_small_sample(small_path, feature_cols, MAX_ROWS)

print("\nLarge dataset sample:", large_sample.shape)
print("Smaller dataset sample:", small_sample.shape)

# --------------------------------------------------
# 3. Compare large up/down moves against small moves
# --------------------------------------------------
large_sample[TARGET] = pd.to_numeric(
    large_sample[TARGET], errors="coerce"
)
small_sample[TARGET] = pd.to_numeric(
    small_sample[TARGET], errors="coerce"
)

large_up = large_sample[large_sample[TARGET] > 500]
large_down = large_sample[large_sample[TARGET] < -500]

def compare_features(group_a, group_b, feature_cols):
    results = []

    for col in feature_cols:
        a = pd.to_numeric(group_a[col], errors="coerce").dropna()
        b = pd.to_numeric(group_b[col], errors="coerce").dropna()

        if len(a) < 20 or len(b) < 20:
            continue

        pooled_sd = np.sqrt(
            (a.var() + b.var()) / 2
        )

        if not np.isfinite(pooled_sd) or pooled_sd == 0:
            continue

        results.append({
            "feature": col,
            "large_move_mean": a.mean(),
            "small_move_mean": b.mean(),
            "standardized_difference": (
                a.mean() - b.mean()
            ) / pooled_sd,
            "absolute_difference": abs(
                (a.mean() - b.mean()) / pooled_sd
            ),
            "large_valid_rows": len(a),
            "small_valid_rows": len(b),
        })

    return pd.DataFrame(results).sort_values(
        "absolute_difference", ascending=False
    )


up_results = compare_features(
    large_up, small_sample, feature_cols
)
down_results = compare_features(
    large_down, small_sample, feature_cols
)

print("\nTOP FEATURES: LARGE UP VS SMALLER MOVES")
display(up_results.head(10))

print("\nTOP FEATURES: LARGE DOWN VS SMALLER MOVES")
display(down_results.head(10))

# --------------------------------------------------
# 4. Plot the strongest differences
# --------------------------------------------------
def plot_top_features(results, title):
    if results.empty:
        print("Not enough valid observations to plot.")
        return

    top = results.head(10).sort_values(
        "absolute_difference"
    )

    plt.figure(figsize=(10, 6))
    plt.barh(
        top["feature"].str.replace("f4z__", "", regex=False),
        top["standardized_difference"]
    )
    plt.axvline(0, linewidth=1)
    plt.xlabel("Standardized difference")
    plt.title(title)
    plt.tight_layout()
    plt.show()


plot_top_features(
    up_results, "Feature differences: Large upward moves"
)
plot_top_features(
    down_results, "Feature differences: Large downward moves"
)

del large_sample, small_sample
gc.collect()
```

```python
import gc
import numpy as np
import pandas as pd
import pyarrow.parquet as pq

# Large-movement dataset only
path = large_path

TARGET = "signed_return"
TIME = "decision_time"
MAX_ROW_GROUPS = 8
BATCH_SIZE = 2048
MAX_SAMPLE_ROWS = 4000

pf = pq.ParquetFile(path)
columns = pf.schema_arrow.names

# Select representative short- and long-window features.
# We will inspect the actual names rather than assume every
# feature uses the same naming convention.
families = [
    "flow", "price", "liquidation", "volatility",
    "oi", "funding", "basis"
]

def window_priority(name):
    n = name.lower()
    if "15m" in n:
        return 0
    if "1h" in n:
        return 1
    if "4h" in n:
        return 2
    if "12h" in n:
        return 3
    return 4

selected = []

for family in families:
    candidates = [
        c for c in columns
        if c.endswith("__value")
        and f"__{family}__" in c.lower()
    ]

    candidates.sort(key=lambda c: (window_priority(c), c))

    # Up to 4 columns per family to keep memory usage low
    selected.extend(candidates[:4])

selected = list(dict.fromkeys(selected))

read_cols = [
    c for c in [TIME, TARGET] + selected
    if c in columns
]

print("Total dataset rows:", f"{pf.metadata.num_rows:,}")
print("Total dataset columns:", len(columns))
print("Selected feature columns:", len(selected))
print("\nSelected features:")
for c in selected:
    print(c)

# Choose a few row groups spread across the file.
# This limits I/O and avoids reading the whole dataset.
num_groups = pf.metadata.num_row_groups
group_ids = np.unique(
    np.linspace(
        0, num_groups - 1,
        min(MAX_ROW_GROUPS, num_groups),
        dtype=int
    )
)

samples = []

for group_id in group_ids:
    for batch in pf.iter_batches(
        batch_size=BATCH_SIZE,
        row_groups=[int(group_id)],
        columns=read_cols
    ):
        frame = batch.to_pandas()

        # Keep only a small sample from each batch.
        take = min(20, len(frame))
        if take:
            samples.append(
                frame.sample(n=take, random_state=int(group_id) + len(samples))
            )

        del frame, batch

data = pd.concat(samples, ignore_index=True) if samples else pd.DataFrame()
del samples
gc.collect()

if not data.empty:
    data = data.drop_duplicates(subset=[TIME, TARGET])
    if len(data) > MAX_SAMPLE_ROWS:
        data = data.sample(
            n=MAX_SAMPLE_ROWS, random_state=42
        ).reset_index(drop=True)

print("\nSample rows retained:", len(data))
print("Sample time range:",
      data[TIME].min(), "to", data[TIME].max())
print("\nReturn summary:")
print(pd.to_numeric(data[TARGET], errors="coerce").describe())
```

```python
import numpy as np
import pandas as pd

rows = []

for col in top_features:
    up = pd.to_numeric(
        positive_df[col], errors="coerce"
    ).dropna()

    down = pd.to_numeric(
        negative_values[col], errors="coerce"
    ).dropna()

    # Pooled standard deviation
    pooled_std = np.sqrt(
        (up.var(ddof=1) + down.var(ddof=1)) / 2
    )

    effect_size = (
        (up.mean() - down.mean()) / pooled_std
        if pooled_std > 0 else np.nan
    )

    rows.append({
        "feature": col,
        "up_count": len(up),
        "up_median": up.median(),
        "up_q25": up.quantile(0.25),
        "up_q75": up.quantile(0.75),
        "down_count": len(down),
        "down_median": down.median(),
        "down_q25": down.quantile(0.25),
        "down_q75": down.quantile(0.75),
        "effect_size": effect_size
    })

effect_table = pd.DataFrame(rows)

print(effect_table.round(4).to_string(index=False))
```

```python
import pandas as pd
import pyarrow.parquet as pq

path = (
    "/content/drive/MyDrive/Scalewave/"
    "BTCUSDT_signed_return_above_500_or_below_minus500_"
    "20261027_093028_e1453352.parquet"
)

pf = pq.ParquetFile(path)
columns = pf.schema.names

time_cols = [
    c for c in columns
    if c.endswith("__available_after")
]

needed = [
    c for c in [
        "decision_time",
        "canonical_symbol",
        "horizon_minutes",
        "direction",
        "signed_return",
    ] + time_cols
    if c in columns
]

df_audit = pd.read_parquet(path, columns=needed)

df_audit["decision_time"] = pd.to_datetime(
    df_audit["decision_time"], utc=True
)

print("TARGET CHECK")
print("Horizons:", df_audit["horizon_minutes"].unique())
print("Symbols:", df_audit["canonical_symbol"].value_counts().to_dict())
print("Direction versus return:")

print(pd.crosstab(
    df_audit["direction"],
    df_audit["signed_return"] > 0,
    rownames=["direction"],
    colnames=["return_positive"]
))

print("\nFEATURE TIMESTAMP CHECK")

for col in time_cols:
    available = pd.to_datetime(
        df_audit[col], utc=True, errors="coerce"
    )

    valid = available.notna()
    lag = (
        available[valid] -
        df_audit.loc[valid, "decision_time"]
    ).dt.total_seconds() / 60

    print(f"\n{col}")
    print("Non-null:", int(valid.sum()))
    print("After decision time:", int((lag > 0).sum()))
    print("Equal to decision time:", int((lag == 0).sum()))
    print("Before decision time:", int((lag < 0).sum()))
    print("Lag in minutes, median:", lag.median())
    print("Lag in minutes, maximum:", lag.max())
```

```python
import pandas as pd
import numpy as np

from sklearn.dummy import DummyClassifier
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    average_precision_score,
    roc_auc_score,
    classification_report,
)

path = (
    "/content/drive/MyDrive/Scalewave/"
    "BTCUSDT_signed_return_above_500_or_below_minus500_"
    "20261027_093028_e1453352.parquet"
)

features = [
    "f4z__flow__flow_activity_percentile__12h__value",
    "f4z__flow__flow_acceleration_percentile__5m__12h__value",
    "f4z__flow__flow_acceleration_percentile__15m__12h__value",
    "f4z__flow__flow_acceleration_percentile__4h__12h__value",
    "f4z__flow__flow_acceleration_percentile__1h__12h__value",
]

target_columns = ["decision_time", "direction"]
df = pd.read_parquet(path, columns=target_columns + features)

df["decision_time"] = pd.to_datetime(
    df["decision_time"], utc=True
)

# 1 = upward outcome, 0 = downward outcome
df["target"] = (df["direction"] == 1).astype(int)

df = df.sort_values("decision_time").reset_index(drop=True)

# Use chronological boundaries; do not randomly shuffle time-series data
times = np.sort(df["decision_time"].unique())
t70 = times[int(len(times) * 0.70)]
t85 = times[int(len(times) * 0.85)]

# Purge 48 hours around each split boundary because target windows
# span 48 hours. This reduces overlap between adjacent partitions.
gap = pd.Timedelta(minutes=2880)

train = df[df["decision_time"] < t70 - gap].copy()
valid = df[
    (df["decision_time"] >= t70 + gap) &
    (df["decision_time"] < t85 - gap)
].copy()
test = df[df["decision_time"] >= t85 + gap].copy()

print("TRAIN:", len(train), train["decision_time"].min(),
      "to", train["decision_time"].max())
print("VALIDATION:", len(valid), valid["decision_time"].min(),
      "to", valid["decision_time"].max())
print("TEST:", len(test), test["decision_time"].min(),
      "to", test["decision_time"].max())

if min(len(train), len(valid), len(test)) == 0:
    raise ValueError("A time split is empty. Inspect date coverage.")

X_train = train[features].replace([np.inf, -np.inf], np.nan)
X_valid = valid[features].replace([np.inf, -np.inf], np.nan)
X_test = test[features].replace([np.inf, -np.inf], np.nan)

y_train = train["target"]
y_valid = valid["target"]
y_test = test["target"]

# Baseline: always predict the most common training class
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)
baseline_pred = dummy.predict(X_test)

# Logistic regression with missing-value handling and scaling
model = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    LogisticRegression(max_iter=1000, class_weight="balanced")
)

model.fit(X_train, y_train)

# Choose a threshold on validation data, not on the test set
valid_prob = model.predict_proba(X_valid)[:, 1]
threshold = 0.50
valid_pred = (valid_prob >= threshold).astype(int)

test_prob = model.predict_proba(X_test)[:, 1]
test_pred = (test_prob >= threshold).astype(int)

print("\nBASELINE — TEST")
print("Accuracy:", round(accuracy_score(y_test, baseline_pred), 4))
print("Balanced accuracy:",
      round(balanced_accuracy_score(y_test, baseline_pred), 4))

print("\nLOGISTIC REGRESSION — VALIDATION")
print("Accuracy:", round(accuracy_score(y_valid, valid_pred), 4))
print("Balanced accuracy:",
      round(balanced_accuracy_score(y_valid, valid_pred), 4))
print("Average precision:",
      round(average_precision_score(y_valid, valid_prob), 4))
print("ROC-AUC:", round(roc_auc_score(y_valid, valid_prob), 4))

print("\nLOGISTIC REGRESSION — TEST")
print("Accuracy:", round(accuracy_score(y_test, test_pred), 4))
print("Balanced accuracy:",
      round(balanced_accuracy_score(y_test, test_pred), 4))
print("Average precision:",
      round(average_precision_score(y_test, test_prob), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, test_prob), 4))
print("\nClassification report:")
print(classification_report(
    y_test, test_pred,
    target_names=["Downward", "Upward"],
    zero_division=0
))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Saved notebook averages—not recalculated from the raw datasets.
data = pd.DataFrame([
    {
        "column": "f4z__basis__basis_persistence_percentile__12h__value",
        "label": "Basis persistence · 12h",
        "scale": "percentile",
        "upward": 0.435975,
        "smaller": 0.643485,
    },
    {
        "column": "f4z__liquidation__liq_side_pressure_rz__12h__value",
        "label": "Liquidation-side pressure · 12h",
        "scale": "rz",
        "upward": 0.493981,
        "smaller": -0.025395,
    },
    {
        "column": "f4z__basis__basis_persistence_percentile__1h__value",
        "label": "Basis persistence · 1h",
        "scale": "percentile",
        "upward": 0.428667,
        "smaller": 0.557771,
    },
    {
        "column": "f4z__liquidation__liq_side_pressure_percentile__12h__value",
        "label": "Liquidation-side pressure · 12h",
        "scale": "percentile",
        "upward": 0.669027,
        "smaller": 0.490856,
    },
    {
        "column": "f4z__basis__basis_persistence_percentile__4h__value",
        "label": "Basis persistence · 4h",
        "scale": "percentile",
        "upward": 0.481877,
        "smaller": 0.595201,
    },
    {
        "column": "f4z__flow__flow_cvd_slope_percentile__12h__value",
        "label": "Flow CVD slope · 12h",
        "scale": "percentile",
        "upward": 0.584991,
        "smaller": 0.460658,
    },
    {
        "column": "f4z__flow__flow_pressure_percentile__12h__value",
        "label": "Flow pressure · 12h",
        "scale": "percentile",
        "upward": 0.587632,
        "smaller": 0.467347,
    },
])

fig, axes = plt.subplots(
    2, 1,
    figsize=(13, 10),
    gridspec_kw={"height_ratios": [6, 1.6]},
)
bar_height = 0.34

for ax, scale in zip(axes, ["percentile", "rz"]):
    subset = data[data["scale"] == scale]
    y = np.arange(len(subset))

    up_bars = ax.barh(
        y - bar_height / 2,
        subset["upward"],
        height=bar_height,
        color="#159570",
        label="Upward moves > +500 bps",
    )
    small_bars = ax.barh(
        y + bar_height / 2,
        subset["smaller"],
        height=bar_height,
        color="#8C9BAA",
        label="Smaller moves: −500 to +500 bps",
    )

    ax.bar_label(up_bars, fmt="%.4f", padding=5, fontsize=10)
    ax.bar_label(small_bars, fmt="%.4f", padding=5, fontsize=10)

    ax.set_yticks(y)
    ax.set_yticklabels(subset["label"], fontsize=11)
    ax.invert_yaxis()
    ax.axvline(0, color="#555555", linewidth=0.8)
    ax.grid(axis="x", alpha=0.2)
    ax.set_axisbelow(True)
    ax.spines[["top", "right"]].set_visible(False)

    if scale == "percentile":
        ax.set_xlim(0, 1)
        ax.set_title("Percentile features", loc="left", fontsize=12)
        ax.set_xlabel("Average feature value (0–1 scale)")
    else:
        ax.set_xlim(-0.15, 0.75)
        ax.set_title("RZ feature — separate scale", loc="left", fontsize=12)
        ax.set_xlabel("Average feature value (rz scale)")

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles, labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.945),
    ncol=2,
    frameon=False,
)

fig.suptitle(
    "Bitcoin: Feature Averages for Upward Moves > +500 bps\n"
    "48-hour return outcome",
    fontsize=17,
    fontweight="bold",
    y=0.995,
)

fig.text(
    0.5, 0.015,
    "Source: saved exploratory notebook results. "
    "Group averages show associations, not causal effects or prediction thresholds.",
    ha="center",
    fontsize=10,
    color="#555555",
)

plt.tight_layout(rect=[0, 0.045, 1, 0.90])
plt.savefig("upward_500bps_feature_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

# Export the exact feature names and values used in the graph.
data.to_csv("upward_500bps_graph_data.csv", index=False)
```
